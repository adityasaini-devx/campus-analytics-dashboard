# 01. Exploratory Data Analysis (EDA) - CampusPulse BIComprehensive exploration of the 8,426 student campus dataset across 6 academic departments.We analyze feature distributions, correlations, and relationships with the academic risk target.

```python
import pandas as pdimport numpy as npimport matplotlib.pyplot as pltimport seaborn as snssns.set_theme(style='whitegrid', palette='muted')df = pd.read_csv('../data/raw/campus_students_8426.csv')print(f'Dataset Shape: {df.shape}')df.head()
```

## 1. Dataset Overview & Data Types

```python
print(df.info())print('\nMissing values per column:')print(df.isnull().sum())print(f'\nDuplicate student rows: {df.duplicated().sum()}')
```

## 2. Statistical Analysis & Central Tendencies

```python
df.describe().T
```

## 3. Key Feature DistributionsExamining attendance percentage, current CGPA, internal marks, and active backlogs.

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))sns.histplot(df['attendance_percentage'], kde=True, ax=axes[0, 0], color='#4f46e5')axes[0, 0].axvline(75.0, color='red', linestyle='--', label='75% Statutory Cutoff')axes[0, 0].set_title('Attendance Distribution')axes[0, 0].legend()sns.histplot(df['current_cgpa'], kde=True, ax=axes[0, 1], color='#0ea5e9')axes[0, 1].set_title('Cumulative CGPA Distribution')sns.countplot(data=df, x='backlogs', ax=axes[1, 0], palette='Blues')axes[1, 0].set_title('Course Backlog Distribution')sns.histplot(df['event_participation'], kde=True, ax=axes[1, 1], color='#10b981')axes[1, 1].set_title('Campus Engagement Score Distribution')plt.tight_layout()plt.show()
```

## 4. Academic Relationships & Risk Drivers

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))sns.scatterplot(data=df, x='attendance_percentage', y='current_cgpa', hue='at_risk', alpha=0.6, ax=axes[0], palette={0: '#10b981', 1: '#ef4444'})axes[0].set_title('Attendance vs CGPA by Risk Status')sns.boxplot(data=df, x='at_risk', y='current_cgpa', ax=axes[1], palette=['#10b981', '#ef4444'])axes[1].set_title('CGPA by Risk Class')sns.barplot(data=df, x='department', y='at_risk', ax=axes[2], palette='viridis')axes[2].set_title('At-Risk Rate by Department')plt.tight_layout()plt.show()
```

## 5. Summary Insights- Attendance displays a normal distribution centered at ~81.4%.- Backlogs and acute attendance deficits (<75%) exhibit strong non-linear separation with the `at_risk` target.
